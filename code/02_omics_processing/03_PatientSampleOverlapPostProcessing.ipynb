```python
# Imports
import pandas as pd
import numpy as np
import pickle
from pathlib import Path

# ----------------
# Configuration
# ----------------
datadir = Path("/work/gr-fe/bryan/data/SHCS/")  # adjust if needed
cohort = ""
omics = ["PGS" , "Metabolomics", "Proteomics"]

input_dir = datadir / cohort / "02_processed"
output_dir = input_dir  # save alongside inputs
output_dir.mkdir(parents=True, exist_ok=True)

print(f"Input dir:  {input_dir}")
print(f"Output dir: {output_dir}")


idx_list = []
# ----------------
# Processing loop
# ----------------
for omic in omics:
    print(f"\n=== Processing {omic} ===")

    # Load
    in_path = input_dir / f"{omic}.processed.pkl"
    with open(in_path, "rb") as f:
        data = pickle.load(f)
    print(f" - Laoded: {in_path}")

    idx_list = np.unique(np.append(idx_list , data['meta'].index.values))

# Load data
print(f"\n=== Loading Phenotype File ===")
with open(f"{input_dir}/phenotype.processed.pkl", "rb") as file:
    meta = pickle.load(file)

print(f"\n=== Subset to Omic IDs ===")
meta = meta[meta['ID'].isin(idx_list)]

# Save
out_path = output_dir / f"phenotype.processed.pkl"
with open(out_path, "wb") as f:
    pickle.dump(meta, f)
print(f" - Saved: {out_path}")
```
