# MOGSage Hyperparameter Grid Search
Generate a text file of `python mogsage.py ...` commands — one per hyperparameter combination.  
Edit the `GRID` dict in the next cell, then **Run All**.

```python
import itertools
```

## 1. Define the hyperparameter grid

```python
GRID = {
    # --- Architecture ---
    "hidden_dim":     [64, 128],
    "emb_dim":        [32, 64],
    "decoder_dim":    [64, 128],

    # --- Omics ---
    "enc_dims" : ["128 128 128", "64 64 64"],
    "latent_dims" : ["64 64 64", "32 32 32"],

    # --- Contrastive / distillation ---
    "alpha_contrast": [0.1, 0.2, 0.5],
    "temperature":    [0.07],

    # --- Optimiser ---
    "lr":             [1e-4, 5e-4, 1e-3],
    "weight_decay":   [1e-4],

    # --- Regularisation ---
    "dropout":        [0.3, 0.5],
}

# Script to invoke
SCRIPT     = "mogsage.py"
OUTPUT_FILE = "./../../hpc_deployment/gridsearch.txt"

print(f"Grid keys  : {list(GRID.keys())}")
print(f"Total combos: {len(list(itertools.product(*GRID.values())))}")
```

## 2. Generate commands

```python
def generate_commands(grid: dict, script: str = "mogsage.py") -> list[str]:
    """Return one CLI command string per hyperparameter combination."""
    keys   = list(grid.keys())
    values = list(grid.values())
    commands = []
    for combo in itertools.product(*values):
        args = " ".join(f"--{k} {v}" for k, v in zip(keys, combo))
        commands.append(f"python {script} {args}")
    return commands

commands = generate_commands(GRID, script=SCRIPT)
print(f"Generated {len(commands)} commands")
print("\nFirst 3 commands (preview):")
for cmd in commands[:3]:
    print(" ", cmd)
```

## 3. Write to file

```python
with open(OUTPUT_FILE, "w") as f:
    f.write("\n".join(commands) + "\n")

print(f"Wrote {len(commands)} commands to '{OUTPUT_FILE}'")
```

## 4. Preview output file (optional)

```python
N_PREVIEW = 10  # change to see more lines

with open(OUTPUT_FILE) as f:
    lines = f.readlines()

print(f"Total lines: {len(lines)}\n")
for line in lines[:N_PREVIEW]:
    print(line, end="")
if len(lines) > N_PREVIEW:
    print(f"\n... and {len(lines) - N_PREVIEW} more.")
```
